```python
df["TotalCharges"] = df["TotalCharges"].fillna(df["TotalCharges"].median())
```

```python
df.columns = df.columns.str.lower()
```

```python
df["age"] = df["age"].astype(int)
df["monthlycharges"] = df["monthlycharges"].astype(float)
df["totalcharges"] = df["totalcharges"].astype(float)
```

```python
df["margin"] = df["totalcharges"] - df["monthlycharges"]
```

```python
df["high_value_customer"] = df["monthlycharges"] > 80
```

```python
customers = df[[
    "customerid",
    "age",
    "gender",
    "tenure"
]]
```

```python
churn = df[[
    "customerid",
    "monthlycharges",
    "totalcharges",
    "churn"
]]
```

```python
import os

# Create the 'output' directory if it doesn't exist
os.makedirs('output', exist_ok=True)

customers.to_csv("output/customers.csv", index=False)
churn.to_csv("output/churn_summary.csv", index=False)
```

Customer Churn ETL Pipeline

Steps Performed:

1. Loaded raw dataset from /raw folder.
2. Cleaned missing values using median imputation.
3. Removed duplicate records.
4. Standardized column names to lowercase.
5. Converted datatypes for numerical fields.
6. Created derived columns (margin, high_value_customer).
7. Split dataset into customers and churn tables.
8. Exported results to CSV and SQLite database.
9. Validated row counts to ensure ETL accuracy.

Outcome:
Clean and structured data ready for analytics and dashboards.
